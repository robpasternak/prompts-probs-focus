```python
!source ../.venv/bin/activate

import pandas as pd
from lmfocus.utils import (
    RANDOM_FOCUS_MARKERS,
    GEMMA_MODELS,
    QWEN_3_MODELS,
    QWEN_35_MODELS,
)
from itertools import product

stats_folder = "../stats/"
```

```python
df_1 = pd.read_csv(stats_folder + 'stats_exp_01.csv')
df_2 = pd.read_csv(stats_folder + 'stats_exp_02.csv')
```

```python
def accumulate_stats(df):
    df_prompt_better_sd = df[(df['term'] != '(Intercept)') & (df['estimate'] > 0) & (df['p_value'] < .05)]
    df_prompt_better_marg = df[(df['term'] != '(Intercept)') & (df['estimate'] > 0) & (df['p_value'] >= .05) & (df['p_value'] < .1)]
    df_prompt_worse_sd = df[(df['term'] != '(Intercept)') & (df['estimate'] < 0) & (df['p_value'] < .05)]
    df_prompt_worse_marg = df[(df['term'] != '(Intercept)') & (df['estimate'] < 0) & (df['p_value'] >= .05) & (df['p_value'] < .1)]
    df_prompt_nsd = df[(df['term'] != '(Intercept)') & (df['estimate'] < 0) & (df['p_value'] >= .1)]
    return (
        df_prompt_better_sd,
        df_prompt_better_marg,
        df_prompt_worse_sd,
        df_prompt_worse_marg,
        df_prompt_nsd
    )
```

# Experiment 1

```python
(
    df1_prompt_better_sd,
    df1_prompt_better_marg,
    df1_prompt_worse_sd,
    df1_prompt_worse_marg,
    df1_prompt_nsd
) = accumulate_stats(df_1)
```
